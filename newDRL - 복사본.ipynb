```python
#학습 결과 집계 스크립트(성공률 등)
import pandas as pd
from pathlib import Path
import os, glob, datetime  #에피소드 총개수가 1 더 많음 > 성공+실패

EXP_DIR = "/home/sooyeon/ray_results/lazy_initial_test_030424"
DATE_FORMAT = "%Y-%m-%d_%H-%M-%S"  # progress.csv date 형식
EPIS_COLS  = ["sampler_results/episodes_this_iter", "episodes_this_iter"]
SUCC_COLS  = ["custom_metrics/success_mean", "sampler_results/custom_metrics/success_mean"]
FAIL_COLS  = ["custom_metrics/failure_mean", "sampler_results/custom_metrics/failure_mean"]
TO_COLS    = ["custom_metrics/is_timeout_mean", "sampler_results/custom_metrics/is_timeout_mean"]
CL_COLS    = ["custom_metrics/is_comm_loss_mean", "sampler_results/custom_metrics/is_comm_loss_mean"]

def pick_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

def iter_count(df, mean_col, ep_col):
    if mean_col is None or ep_col is None:
        return 0.0
    x = df[[mean_col, ep_col]].dropna()
    if x.empty:
        return 0.0
    return float((x[mean_col] * x[ep_col]).sum())

# 집계 구간 정의
ranges = [
    #("2025-08-10", "2025-08-11 12:00"),
    #("2025-08-11 12:00", "2025-08-12 15:59:59"),
    #("2025-08-12 16:00:00", "2025-08-13 23:59:59"),
    #("2025-08-19", "2025-08-21 23:59:59"),
    #("2025-09-02", "2025-09-03 23:59:59"),
    ("2025-11-04 16:00", "2025-11-05 14:30"), #커넥티드
    ("2025-10-30 19:00", "2025-10-31 23:59:59"), # 디스커넥티드
    #("2025-11-01", "2025-11-02 23:59:59"), #디스커넥티드 150
    #("2025-11-03 11:00 ", "2025-11-03 23:59:59"), #커넥티드 150
 ]

for start_str, end_str in ranges:
    start_dt = pd.to_datetime(start_str)
    end_dt = pd.to_datetime(end_str)

    tot_succ = tot_fail = tot_to = tot_cl = 0
    trials = 0

    for prog in glob.glob(os.path.join(EXP_DIR, "**", "progress.csv"), recursive=True):
        trial_dir = Path(prog).parent
        df = pd.read_csv(prog)
        if df.empty or "date" not in df.columns:
            continue

        dts = pd.to_datetime(df["date"], format=DATE_FORMAT, errors="coerce")
        mask = (dts >= start_dt) & (dts <= end_dt)
        df_range = df.loc[mask]
        if df_range.empty:
            continue

        ep_col   = pick_col(df_range, EPIS_COLS)
        succ_col = pick_col(df_range, SUCC_COLS)
        fail_col = pick_col(df_range, FAIL_COLS)
        to_col   = pick_col(df_range, TO_COLS)
        cl_col   = pick_col(df_range, CL_COLS)

        n_succ = iter_count(df_range, succ_col, ep_col)
        n_fail = iter_count(df_range, fail_col, ep_col)
        n_to   = iter_count(df_range, to_col,   ep_col) if to_col else 0.0
        n_cl   = iter_count(df_range, cl_col,   ep_col) if cl_col else 0.0

        tot_succ += n_succ
        tot_fail += n_fail
        tot_to   += n_to
        tot_cl   += n_cl
        trials += 1

    total_eps = tot_succ + tot_fail
    sr = (tot_succ / total_eps) if total_eps > 0 else 0.0
    fr = (tot_fail / total_eps) if total_eps > 0 else 0.0
    to_fail_ratio = (tot_to / tot_fail) if tot_fail > 0 else 0.0
    cl_fail_ratio = (tot_cl / tot_fail) if tot_fail > 0 else 0.0

    print(f"\n=== {start_dt} ~ {end_dt} ===")
    print(f"참여 트라이얼: {trials}")
    print(f"총 Episodes: {int(total_eps)}")
    print(f"성공: {int(tot_succ)}, 실패: {int(tot_fail)}")
    print(f"성공률: {sr:.2%}, 실패률: {fr:.2%}")
    print(f"타임아웃: {int(tot_to)}, 통신단절: {int(tot_cl)}")
    print(f"실패 중 타임아웃 비율: {to_fail_ratio:.2%}")
    print(f"실패 중 통신단절 비율: {cl_fail_ratio:.2%}")
```

```python
#vx, vy 그래프 저장 스크립트
import numpy as np
# matplotlib
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib import cm
# Save files and load files
import pickle
import os  # creates dirs
from datetime import datetime  # gets current date and time
from ray.tune.logger import pretty_print

max_time_steps = 1000

def get_save_path_and_file_name(date_str: str, start_seed: int, last_seed: int):
    # Get path
    base_path = "/home/sooyeon/data/"
    dir_path = os.path.join(base_path, date_str + "/")
    #
    seed_range_str = f"seed_{start_seed}-{last_seed}_"
    plk_name = "001_acs_vs_rl_" + seed_range_str + date_str + ".pkl"

    return dir_path, plk_name

# Get path
# folder_path, file_name = get_save_path_and_file_name("2024-03-12_14-13-31", 0, 9)
folder_path, file_name = get_save_path_and_file_name("2025-11-04_03-50-26", 20, 29)

 # Load and check
with open(os.path.join(folder_path, file_name), "rb") as f:
    data_loaded = pickle.load(f)

# Assign explicit variables
trajectories = data_loaded['trajectories']  # (num_seeds, num_algos, max_time_steps, num_agents, 2)
velocities = data_loaded['velocities'][:, :, :max_time_steps, :, :]  # 🔥 타임스텝까지만 가져오기
algo_str = data_loaded['algo_str']  # (num_algos)
seeds = data_loaded['seeds']        # (num_seeds)

# Dimensions
num_seeds = len(seeds)
num_algos = len(algo_str)
#max_time_steps = velocities.shape[2]
num_agents = trajectories.shape[3]



# 저장할 폴더 생성
if not os.path.exists(folder_path + "plots1/"):
    os.makedirs(folder_path + "plots1/")

# 각 시드(episode)에 대해 그래프 저장
for episode in range(num_seeds):
    fig, axes = plt.subplots(2, num_algos, figsize=(15, 10), constrained_layout=True)

    for algo_index in range(num_algos):
        # vx, vy 데이터 추출
        vx = velocities[episode, algo_index, :max_time_steps, :, 0]  # vx 값
        vy = velocities[episode, algo_index, :max_time_steps, :, 1]  # vy 값

        # 모든 에이전트의 vx 플롯
        for agent_idx in range(num_agents):
            axes[0, algo_index].plot(np.arange(max_time_steps), vx[:, agent_idx], alpha=0.7)

        axes[0, algo_index].set_title(f"vx - {algo_str[algo_index]} (Seed {seeds[episode]})")
        axes[0, algo_index].set_xlabel("Time Steps")
        axes[0, algo_index].set_ylabel("vx")
        axes[0, algo_index].grid(True, linestyle="--", linewidth=0.5, alpha=0.67)

        # 모든 에이전트의 vy 플롯
        for agent_idx in range(num_agents):
            axes[1, algo_index].plot(np.arange(max_time_steps), vy[:, agent_idx], alpha=0.7)

        axes[1, algo_index].set_title(f"vy - {algo_str[algo_index]} (Seed {seeds[episode]})")
        axes[1, algo_index].set_xlabel("Time Steps")
        axes[1, algo_index].set_ylabel("vy")
        axes[1, algo_index].grid(True, linestyle="--", linewidth=0.5, alpha=0.67)

    # 그래프 저장 (시드 번호 포함)
    save_path = folder_path + f"plots1/velocity_seed_{seeds[episode]}.png"
    plt.savefig(save_path, dpi=300)
    plt.close()
    print(f"Velocity vx, vy plot for seed {seeds[episode]} saved at {save_path}")

print("Velocity vx, vy plots for all seeds generated successfully!")
```
