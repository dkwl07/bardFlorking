```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# 에피소드 개수 (원하면 50 말고 다른 값으로 바꿔도 됨)
N_EPISODES = 50

for ep in range(1, N_EPISODES + 1):
    csv_path = f"/home/sooyeon/ray_results/discrete/connected_constraint(300s,train)/test_results/velocities_episode_{ep:03d}.csv"

    if not os.path.exists(csv_path):
        print(f"[경고] {csv_path} 파일이 없어서 스킵합니다.")
        continue

    # === 1. CSV 불러오기 ===
    df = pd.read_csv(csv_path)

    # 혹시 다른 에피소드가 섞여있을 수도 있으니 한번 더 필터
    if "episode_idx" in df.columns:
        df = df[df["episode_idx"] == ep]

    if df.empty:
        print(f"[경고] {csv_path} 내용이 비어있어서 스킵합니다.")
        continue

    agent_ids = sorted(df["agent_id"].unique())

    # === 2. 한 이미지 안에 vx / vy 두 개 subplot ===
    fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

    # --- 위쪽: vx ---
    for agent_id in agent_ids:
        sub = df[df["agent_id"] == agent_id]
        axes[0].plot(sub["step"], sub["vx"], alpha=0.4)
    axes[0].set_ylabel("vx")
    axes[0].set_title(f"Episode {ep}: vx of all agents over time")
    axes[0].grid(True)

    # --- 아래쪽: vy ---
    for agent_id in agent_ids:
        sub = df[df["agent_id"] == agent_id]
        axes[1].plot(sub["step"], sub["vy"], alpha=0.4)
    axes[1].set_xlabel("Timestep")
    axes[1].set_ylabel("vy")
    axes[1].set_title(f"Episode {ep}: vy of all agents over time")
    axes[1].grid(True)

    plt.tight_layout()

    # === 3. PNG로 저장 (CSV와 같은 디렉토리) ===
    img_path = os.path.splitext(csv_path)[0] + ".png"  # velocities_episode_XXX.png
    plt.savefig(img_path, dpi=300)
    plt.close(fig)

    print(f"[완료] 그래프 저장: {img_path}")

print("모든 에피소드 처리 완료!")
```

```python
import os, glob, re
import pandas as pd
from pathlib import Path

# ===== 설정 =====
EXP_DIR   = "/home/sooyeon/ray_results/lazy_initial_test_030424"
DATE_STR  = "2025-08-11"                  # 보고 싶은 날짜 (YYYY-MM-DD)
DATE_FMT  = "%Y-%m-%d_%H-%M-%S"           # progress.csv의 date 형식 (예: 2025-08-11_13-01-41)

# 컬럼 후보들 (Ray 버전/로그 표기 차이 커버)
EPIS_COLS = ["sampler_results/episodes_this_iter", "episodes_this_iter"]
SUCC_COLS = ["custom_metrics/success_mean", "sampler_results/custom_metrics/success_mean"]
FAIL_COLS = ["custom_metrics/failure_mean", "sampler_results/custom_metrics/failure_mean"]
TO_COLS   = ["custom_metrics/is_timeout_mean", "sampler_results/custom_metrics/is_timeout_mean"]
CL_COLS   = ["custom_metrics/is_comm_loss_mean", "sampler_results/custom_metrics/is_comm_loss_mean"]

# ── 유틸 ─────────────────────────────────────────────────────────────────────
def pick_col(df: pd.DataFrame, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

def counts_from_subset(df_day: pd.DataFrame, ep_col: str, mean_col: str) -> float:
    """누적 개수 복원 = sum(episodes_this_iter × mean) (subset에 대해서만)"""
    if ep_col is None or mean_col is None or mean_col not in df_day.columns:
        return 0.0
    x = df_day[[ep_col, mean_col]].dropna()
    if x.empty:
        return 0.0
    return float((x[ep_col] * x[mean_col]).sum())

def trial_starts_on_date(dir_name: str, date_str: str) -> bool:
    """
    트라이얼 폴더명 끝부분 패턴:
    ..._YYYY-MM-DD_HH-MM-SS
    """
    m = re.search(r"_(\d{4}-\d{2}-\d{2})_\d{2}-\d{2}-\d{2}$", dir_name)
    return (m is not None) and (m.group(1) == date_str)

# ── 집계 ─────────────────────────────────────────────────────────────────────
total_success = 0.0
total_failure = 0.0
total_timeout = 0.0
total_closs   = 0.0
total_episodes = 0.0
num_trials_used = 0

for prog in glob.glob(os.path.join(EXP_DIR, "**", "progress.csv"), recursive=True):
    trial_dir = Path(prog).parent

    # 1) 해당 날짜에 "시작한" 트라이얼만 포함
    if not trial_starts_on_date(trial_dir.name, DATE_STR):
        continue

    # 2) CSV 로드 & 날짜 부분집합 만들기 (해당 날짜 행만)
    try:
        df = pd.read_csv(prog)
        if df.empty or "date" not in df.columns:
            continue

        d = pd.to_datetime(df["date"], format=DATE_FMT, errors="coerce")
        if d.isna().all():  # 형식이 다를 경우 폴백
            d = pd.to_datetime(df["date"], errors="coerce")
        df_day = df[d.dt.strftime("%Y-%m-%d") == DATE_STR].copy()
        if df_day.empty:
            continue
    except Exception:
        continue

    # 3) 필요한 컬럼 결정
    ep_col   = pick_col(df_day, EPIS_COLS)
    succ_col = pick_col(df_day, SUCC_COLS)
    fail_col = pick_col(df_day, FAIL_COLS)
    to_col   = pick_col(df_day, TO_COLS)
    cl_col   = pick_col(df_day, CL_COLS)
    if ep_col is None:
        continue

    # 4) 날짜 부분집합으로 누적 개수 복원
    n_succ = counts_from_subset(df_day, ep_col, succ_col)
    n_fail = counts_from_subset(df_day, ep_col, fail_col)
    n_to   = counts_from_subset(df_day, ep_col, to_col)
    n_cl   = counts_from_subset(df_day, ep_col, cl_col)

    # 5) 날짜 부분집합의 에피소드 수
    tot_eps_trial = float(df_day[ep_col].fillna(0).sum())

    # 6) 총합 집계
    total_success  += n_succ
    total_failure  += n_fail
    total_timeout  += n_to
    total_closs    += n_cl
    total_episodes += tot_eps_trial
    num_trials_used += 1

# 보기 좋게 정수화
i = lambda x: int(round(x))

print(f"===== {DATE_STR} 전체 합계 (해당 날짜에 시작한 트라이얼만, 그 날짜 행만 집계) =====")
print(f"참여 트라이얼 수: {num_trials_used}")
print(f"총 에피소드: {i(total_episodes)}")

if total_episodes > 0:
    tot_sf = total_success + total_failure
    succ_rate = (total_success / tot_sf) if tot_sf > 0 else 0.0
    fail_rate = 1.0 - succ_rate

    print(f"성공: {i(total_success)}회 ({succ_rate:.2%})")
    print(f"실패: {i(total_failure)}회 ({fail_rate:.2%})")

    if total_failure > 0:
        to_rate = total_timeout / total_failure
        cl_rate = total_closs   / total_failure
    else:
        to_rate = cl_rate = 0.0

    print(f" - 타임아웃: {i(total_timeout)}회 ({to_rate:.2%})")
    print(f" - 통신단절: {i(total_closs)}회 ({cl_rate:.2%})")
else:
    print("해당 날짜에 집계할 에피소드가 없습니다.")

# 8/11에 시작한 트라이얼 중, 가장 최근(이름순으로 마지막) 하나의 "마지막 줄" 스냅샷도 같이 보기
import re, os
from pathlib import Path
candidates = []
for prog in glob.glob(os.path.join(EXP_DIR, "**", "progress.csv"), recursive=True):
    trial_dir = Path(prog).parent
    m = re.search(r"_(\d{4}-\d{2}-\d{2})_\d{2}-\d{2}-\d{2}$", trial_dir.name)
    if m and m.group(1) == DATE_STR:
        candidates.append((trial_dir.name, prog))

if candidates:
    candidates.sort()
    _, latest_prog = candidates[-1]
    df_latest = pd.read_csv(latest_prog)
    # 마지막 한 줄(스냅샷)
    last = df_latest.sort_values("training_iteration").iloc[-1] \
           if "training_iteration" in df_latest.columns else df_latest.iloc[-1]
    s_last = float(last.get("custom_metrics/success_mean", 0.0))
    f_last = float(last.get("custom_metrics/failure_mean", 0.0))
    to_last = float(last.get("custom_metrics/is_timeout_mean", 0.0))
    cl_last = float(last.get("custom_metrics/is_comm_loss_mean", 0.0))
    print(f"\n===== {DATE_STR} 최신 트라이얼 '마지막 iteration' 스냅샷 =====")
    print(f"최근 성공률: {s_last:.2%}, 최근 실패률: {f_last:.2%}")
    print(f" - 최근 타임아웃률: {to_last:.2%}, 최근 통신단절률: {cl_last:.2%}")
```

```python
import os, glob, re
import pandas as pd
from pathlib import Path

# ===== 설정 =====
EXP_DIR    = "/home/sooyeon/ray_results/lazy_initial_test_030424"
START_DATES = ["2025-08-10", "2025-08-11"]   # 시작일들 (YYYY-MM-DD)
DATE_FMT   = "%Y-%m-%d_%H-%M-%S"             # progress.csv 의 date 포맷 (참고용, 여기선 전체 기간 집계)

# 컬럼 후보 (버전/표기 차이 흡수)
EPIS_COLS = ["sampler_results/episodes_this_iter", "episodes_this_iter"]
SUCC_COLS = ["custom_metrics/success_mean", "sampler_results/custom_metrics/success_mean"]
FAIL_COLS = ["custom_metrics/failure_mean", "sampler_results/custom_metrics/failure_mean"]
TO_COLS   = ["custom_metrics/is_timeout_mean", "sampler_results/custom_metrics/is_timeout_mean"]
CL_COLS   = ["custom_metrics/is_comm_loss_mean", "sampler_results/custom_metrics/is_comm_loss_mean"]

def pick_col(df, cands):
    for c in cands:
        if c in df.columns:
            return c
    return None

def trial_starts_on_date(dir_name: str, date_str: str) -> bool:
    """
    trial 폴더명 끝 패턴: ..._YYYY-MM-DD_HH-MM-SS
    → 여기서 YYYY-MM-DD가 date_str와 같으면 해당 시작일의 트라이얼로 분류
    """
    m = re.search(r"_(\d{4}-\d{2}-\d{2})_\d{2}-\d{2}-\d{2}$", dir_name)
    return (m is not None) and (m.group(1) == date_str)

def aggregate_by_start_date(exp_dir: str, start_date: str):
    total_eps = 0.0
    total_succ = 0.0
    total_fail = 0.0
    total_to   = 0.0
    total_cl   = 0.0
    num_trials = 0

    per_trial_rows = []  # 필요하면 개별 트라이얼도 보고 싶을 때 사용

    for prog in glob.glob(os.path.join(exp_dir, "**", "progress.csv"), recursive=True):
        trial_dir = Path(prog).parent
        if not trial_starts_on_date(trial_dir.name, start_date):
            continue

        try:
            df = pd.read_csv(prog)
        except Exception:
            continue
        if df.empty:
            continue

        ep_col   = pick_col(df, EPIS_COLS)
        succ_col = pick_col(df, SUCC_COLS)
        fail_col = pick_col(df, FAIL_COLS)
        to_col   = pick_col(df, TO_COLS)
        cl_col   = pick_col(df, CL_COLS)
        if ep_col is None:
            continue

        eps_sum = float(df[ep_col].fillna(0).sum())
        succ_sum = float((df.get(succ_col, 0).fillna(0) * df[ep_col].fillna(0)).sum()) if succ_col else 0.0
        fail_sum = float((df.get(fail_col, 0).fillna(0) * df[ep_col].fillna(0)).sum()) if fail_col else 0.0
        to_sum   = float((df.get(to_col,   0).fillna(0) * df[ep_col].fillna(0)).sum()) if to_col   else 0.0
        cl_sum   = float((df.get(cl_col,   0).fillna(0) * df[ep_col].fillna(0)).sum()) if cl_col   else 0.0

        total_eps  += eps_sum
        total_succ += succ_sum
        total_fail += fail_sum
        total_to   += to_sum
        total_cl   += cl_sum
        num_trials += 1

        per_trial_rows.append({
            "trial": trial_dir.name,
            "총 episode": int(round(eps_sum)),
            "성공(복원)": int(round(succ_sum)),
            "실패(복원)": int(round(fail_sum)),
            "타임아웃(복원)": int(round(to_sum)) if to_col else None,
            "통신단절(복원)": int(round(cl_sum)) if cl_col else None,
        })

    # 총합/비율 계산
    i = lambda x: int(round(x))
    sf = total_succ + total_fail
    succ_rate = (total_succ / sf) if sf > 0 else 0.0
    fail_rate = 1.0 - succ_rate if sf > 0 else 0.0
    to_rate = (total_to / total_fail) if total_fail > 0 else 0.0
    cl_rate = (total_cl / total_fail) if total_fail > 0 else 0.0

    summary = {
        "시작일": start_date,
        "참여 트라이얼 수": num_trials,
        "총 episode": i(total_eps),
        "성공(복원)": i(total_succ),
        "실패(복원)": i(total_fail),
        "성공 비율": succ_rate,
        "실패 비율": fail_rate,
        "타임아웃(복원)": i(total_to),
        "통신단절(복원)": i(total_cl),
        "실패 중 타임아웃 비율": to_rate,
        "실패 중 통신단절 비율": cl_rate,
    }
    return summary, per_trial_rows

# ── 실행 및 출력 ──
all_summaries = []
for sd in START_DATES:
    s, rows = aggregate_by_start_date(EXP_DIR, sd)
    all_summaries.append(s)

# 표 형태로 비교 출력
cols = ["시작일", "참여 트라이얼 수", "총 episode", "성공(복원)", "실패(복원)",
        "성공 비율", "실패 비율", "타임아웃(복원)", "통신단절(복원)",
        "실패 중 타임아웃 비율", "실패 중 통신단절 비율"]
df_out = pd.DataFrame(all_summaries)[cols]
print(df_out.to_string(index=False))
```

```python
import pandas as pd
from pathlib import Path
import os, glob, datetime  #에피소드 총개수가 1 더 많음 > 성공+실패

EXP_DIR = "/home/sooyeon/ray_results/lazy_initial_test_030424"
DATE_FORMAT = "%Y-%m-%d_%H-%M-%S"  # progress.csv date 형식
EPIS_COLS  = ["sampler_results/episodes_this_iter", "episodes_this_iter"]
SUCC_COLS  = ["custom_metrics/success_mean", "sampler_results/custom_metrics/success_mean"]
FAIL_COLS  = ["custom_metrics/failure_mean", "sampler_results/custom_metrics/failure_mean"]
TO_COLS    = ["custom_metrics/is_timeout_mean", "sampler_results/custom_metrics/is_timeout_mean"]
CL_COLS    = ["custom_metrics/is_comm_loss_mean", "sampler_results/custom_metrics/is_comm_loss_mean"]

def pick_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

def iter_count(df, mean_col, ep_col):
    if mean_col is None or ep_col is None:
        return 0.0
    x = df[[mean_col, ep_col]].dropna()
    if x.empty:
        return 0.0
    return float((x[mean_col] * x[ep_col]).sum())

# 집계 구간 정의
ranges = [
    #("2025-08-10", "2025-08-11 12:00"),
    #("2025-08-11 12:00", "2025-08-12 15:59:59"),
    #("2025-08-12 16:00:00", "2025-08-13 23:59:59"),
    ##("2025-08-12 16:00:00", "2025-08-13 23:59:59"),
    ##("2025-09-02", "2025-09-03 23:59:59"),
    ##("2025-11-04 16:00", "2025-11-05 14:30"), #커넥티드
    ##("2025-10-30 00:00", "2025-10-30 18:59"), #커넥티드
    ##("2025-10-30 19:00", "2025-10-31 23:59:59"), # 디스커넥티드
    ##("2025-11-01", "2025-11-02 23:59:59"), #디스커넥티드 150
    ##("2025-11-03 11:00 ", "2025-11-03 23:59:59"), #커넥티드 150
    ("2025-11-23 0:00:00", "2025-11-24 08:59:59"), # --- IGNORE ---
 ]

for start_str, end_str in ranges:
    start_dt = pd.to_datetime(start_str)
    end_dt = pd.to_datetime(end_str)

    tot_succ = tot_fail = tot_to = tot_cl = 0
    trials = 0

    for prog in glob.glob(os.path.join(EXP_DIR, "**", "progress.csv"), recursive=True):
        trial_dir = Path(prog).parent
        df = pd.read_csv(prog)
        if df.empty or "date" not in df.columns:
            continue

        dts = pd.to_datetime(df["date"], format=DATE_FORMAT, errors="coerce")
        mask = (dts >= start_dt) & (dts <= end_dt)
        df_range = df.loc[mask]
        if df_range.empty:
            continue

        ep_col   = pick_col(df_range, EPIS_COLS)
        succ_col = pick_col(df_range, SUCC_COLS)
        fail_col = pick_col(df_range, FAIL_COLS)
        to_col   = pick_col(df_range, TO_COLS)
        cl_col   = pick_col(df_range, CL_COLS)

        n_succ = iter_count(df_range, succ_col, ep_col)
        n_fail = iter_count(df_range, fail_col, ep_col)
        n_to   = iter_count(df_range, to_col,   ep_col) if to_col else 0.0
        n_cl   = iter_count(df_range, cl_col,   ep_col) if cl_col else 0.0

        tot_succ += n_succ
        tot_fail += n_fail
        tot_to   += n_to
        tot_cl   += n_cl
        trials += 1

    total_eps = tot_succ + tot_fail
    sr = (tot_succ / total_eps) if total_eps > 0 else 0.0
    fr = (tot_fail / total_eps) if total_eps > 0 else 0.0
    to_fail_ratio = (tot_to / tot_fail) if tot_fail > 0 else 0.0
    cl_fail_ratio = (tot_cl / tot_fail) if tot_fail > 0 else 0.0

    print(f"\n=== {start_dt} ~ {end_dt} ===")
    print(f"참여 트라이얼: {trials}")
    print(f"총 Episodes: {int(total_eps)}")
    print(f"성공: {int(tot_succ)}, 실패: {int(tot_fail)}")
    print(f"성공률: {sr:.2%}, 실패률: {fr:.2%}")
    print(f"타임아웃: {int(tot_to)}, 통신단절: {int(tot_cl)}")
    print(f"실패 중 타임아웃 비율: {to_fail_ratio:.2%}")
    print(f"실패 중 통신단절 비율: {cl_fail_ratio:.2%}")
```

```python
import numpy as np, matplotlib.pyplot as plt, pickle, os, re

# ── 1. 데이터 로드 ─────────────────────────────────────
folder_path = "/home/sooyeon/data/2025-05-31_22-34-54/"
file_name   = "001_acs_vs_rl_seed_20-29_2025-05-31_22-34-54.pkl"
with open(os.path.join(folder_path, file_name), "rb") as f:
    data = pickle.load(f)

connected = data["connected"]          # (seeds, algos, T)
algo_str  = data["algo_str"]           # 예: ["ACS", "RL stochastic", "RL deterministic"]
seeds     = data["seeds"]
num_seeds, num_algos, T = connected.shape

# ── 2. “ACS”와 “RL Stochastic” 인덱스만 추리기 ─────────
wanted_names = ["acs", "rl stochastic"]           # 소문자로 통일
keep_idx = [i for i, name in enumerate(algo_str)
            if name.lower() in wanted_names]

# ── 3. 저장 폴더 ──────────────────────────────────────
save_dir = os.path.join(folder_path, "plots1/")
os.makedirs(save_dir, exist_ok=True)

# ── 4. 시각화 루프 ────────────────────────────────────
for s, seed in enumerate(seeds):
    fig, axes = plt.subplots(1, len(keep_idx),
                             figsize=(4.5 * len(keep_idx), 2.5),
                             constrained_layout=True)

    if len(keep_idx) == 1:
        axes = [axes]                       # 단일 축도 리스트화

    for ax, a in zip(axes, keep_idx):
        conn = connected[s, a]             # True/False 1D
        t    = np.arange(T)

        ax.fill_between(t, 0, 1, where=~conn,
                        color="red", alpha=0.6, step="mid")
        ax.set_ylim(-0.1, 1.1)
        ax.set_yticks([0, 1])
        ax.set_yticklabels(["NC", "C"])
        ax.set_xlabel("Time Step")
        ax.set_title(f"{algo_str[a]} (Seed {seed})")

    fig.suptitle("Connectivity Timeline (Red = Non-Connected)")
    save_path = os.path.join(save_dir, f"connectivity_seed_{seed}.png")
    plt.savefig(save_path, dpi=300)
    plt.close()
    print("Saved:", save_path)

print("선택한 알고리즘(ACS, RL Stochastic)만 시각화 완료!")
```
