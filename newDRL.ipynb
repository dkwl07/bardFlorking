```python
import numpy as np
# matplotlib
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib import cm
# Save files and load files
import pickle
import os  # creates dirs
from datetime import datetime  # gets current date and time
from ray.tune.logger import pretty_print

def get_save_path_and_file_name(date_str: str, start_seed: int, last_seed: int):
    # Get path
    base_path = "/home/sooyeon/data/"
    dir_path = os.path.join(base_path, date_str + "/")
    #
    seed_range_str = f"seed_{start_seed}-{last_seed}_"
    plk_name = "001_acs_vs_rl_" + seed_range_str + date_str + ".pkl"

    return dir_path, plk_name

# Get path
# folder_path, file_name = get_save_path_and_file_name("2024-03-12_14-13-31", 0, 9)
folder_path, file_name = get_save_path_and_file_name("2025-03-26_22-41-34", 20, 29)
# Load and check
with open(os.path.join(folder_path, file_name), "rb") as f:
    data_loaded = pickle.load(f)

algo_str = data_loaded['algo_str']  # (num_algos)
seeds = data_loaded['seeds']        # (num_seeds)
trajectories = data_loaded['trajectories']  # (num_seeds, num_algos, max_time_steps, num_agents, 2)

num_seeds = len(seeds)
num_algos = len(algo_str)
num_agents = trajectories.shape[3]
my_dpi = 300  # no arg: 100 (None), 300 is good for printing

start_t = 0
end_t = 500  # 확대할 시점 구간 설정

# Get time
times = np.arange(end_t)  # 0 ~ 199

for episode in range(num_seeds):
    fig, axes = plt.subplots(1, num_algos, figsize=(15, 5), constrained_layout=True)
    for algo_index in range(num_algos):
        for agent_idx in range(num_agents):
            agent_trajectory = trajectories[episode, algo_index, start_t:end_t, agent_idx, :2]
            points = agent_trajectory.reshape(-1, 1, 2)
            segments = np.concatenate([points[:-1], points[1:]], axis=1)
            norm = plt.Normalize(start_t, end_t)
            lc = LineCollection(segments, cmap='viridis', norm=norm)
            lc.set_array(times[start_t:end_t])
            lc.set_linewidth(1)
            line = axes[algo_index].add_collection(lc)
            axes[algo_index].set_title(algo_str[algo_index])
            axes[algo_index].axis('equal')
    cb = fig.colorbar(line, orientation='vertical')
    cb.set_label(f'Time steps ({start_t} ~ {end_t})')

    if not os.path.exists(folder_path + "plots/"):
        os.makedirs(folder_path + "plots/")
    plt.savefig(folder_path + f"plots/trajectories_zoomed_seed_{seeds[episode]}_{start_t}_{end_t}.png", dpi=my_dpi)
    plt.close()
    print(f"Zoomed trajectory plot {episode+1}/{num_seeds} saved")
```

```python
import numpy as np
# matplotlib
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib import cm
# Save files and load files
import pickle
import os  # creates dirs
from datetime import datetime  # gets current date and time
from ray.tune.logger import pretty_print

def get_save_path_and_file_name(date_str: str, start_seed: int, last_seed: int):
    # Get path
    base_path = "/home/sooyeon/data/"
    dir_path = os.path.join(base_path, date_str + "/")
    #
    seed_range_str = f"seed_{start_seed}-{last_seed}_"
    plk_name = "001_acs_vs_rl_" + seed_range_str + date_str + ".pkl"

    return dir_path, plk_name

# Get path
# folder_path, file_name = get_save_path_and_file_name("2024-03-12_14-13-31", 0, 9)
folder_path, file_name = get_save_path_and_file_name("2025-03-26_22-41-34", 20, 29)
# Load and check
with open(os.path.join(folder_path, file_name), "rb") as f:
    data_loaded = pickle.load(f)

algo_str = data_loaded['algo_str']  # (num_algos)
seeds = data_loaded['seeds']        # (num_seeds)
trajectories = data_loaded['trajectories']  # (num_seeds, num_algos, max_time_steps, num_agents, 2)

num_seeds = len(seeds)
num_algos = len(algo_str)
num_agents = trajectories.shape[3]
my_dpi = 300  # no arg: 100 (None), 300 is good for printing

start_t = 0
end_t = 100  # 확대할 시점 구간 설정

# Get time
times = np.arange(end_t)  # 0 ~ 199

zoom_range = 50  # 확대 범위 (중심 ±60)

for episode in range(num_seeds):
    fig, axes = plt.subplots(1, num_algos, figsize=(15, 5), constrained_layout=True)
    for algo_index in range(num_algos):
        # 초기 프레임에서 에이전트 위치 가져오기 (t = start_t)
        init_positions = trajectories[episode, algo_index, start_t, :, :2]
        center_x, center_y = init_positions.mean(axis=0)  # 중심 계산

        for agent_idx in range(num_agents):
            agent_trajectory = trajectories[episode, algo_index, start_t:end_t, agent_idx, :2]
            points = agent_trajectory.reshape(-1, 1, 2)
            segments = np.concatenate([points[:-1], points[1:]], axis=1)
            norm = plt.Normalize(start_t, end_t)
            lc = LineCollection(segments, cmap='viridis', norm=norm)
            lc.set_array(times[start_t:end_t])
            lc.set_linewidth(1)
            axes[algo_index].add_collection(lc)
            axes[algo_index].set_title(algo_str[algo_index])
            axes[algo_index].axis('equal')

            # ✅ 줌인: 중심 주변만 시야로 설정
            axes[algo_index].set_xlim(center_x - zoom_range, center_x + zoom_range)
            axes[algo_index].set_ylim(center_y - zoom_range, center_y + zoom_range)

    cb = fig.colorbar(lc, orientation='vertical')
    cb.set_label(f'Time steps ({start_t} ~ {end_t})')

    if not os.path.exists(folder_path + "plots/"):
        os.makedirs(folder_path + "plots/")
    plt.savefig(folder_path + f"plots/trajectories_zoomed_center_seed_{seeds[episode]}_{start_t}_{end_t}.png", dpi=my_dpi)
    plt.close()
    print(f"Centered zoom plot {episode+1}/{num_seeds} saved")
```
